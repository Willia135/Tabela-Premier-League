```python
df.groupby('AwayTeam')['FTAG'].sum()
```

```python
gols_casa = df.groupby('HomeTeam')['FTHG'].sum()
gols_fora = df.groupby('AwayTeam')['FTAG'].sum()
gols_fora = gols_fora.rename_axis('HomeTeam')
total_gols = gols_casa + gols_fora
print(total_gols)
```

```python
df[df['FTR'] == 'H'].groupby('HomeTeam').size()
```

```python
df[df['FTR'] == 'A'].groupby('AwayTeam').size()
```

```python
vitorias_casa = df[df['FTR'] == 'H'].groupby('HomeTeam').size()
vitorias_fora = df[df['FTR'] == 'A'].groupby('AwayTeam').size()
```

```python
df[df['FTR'] == 'D'].groupby('HomeTeam').size()
```

```python
df[df['FTR'] == 'D'].groupby('AwayTeam').size()
```

```python
empates_casa = df[df['FTR'] == 'D'].groupby('HomeTeam').size()
empates_fora = df[df['FTR'] == 'D'].groupby('AwayTeam').size()
```

```python
tabela= pd.DataFrame({
    'Vitorias': vitorias_casa.values + vitorias_fora.values,
    'Empates': empates_casa.values + empates_fora.values,
    'Gols': total_gols.values
}, index=total_gols.index)
```

```python
tabela['Pontos'] = tabela['Vitorias'] * 3 + tabela['Empates'] * 1
```

```python
tabela.sort_values(by='Pontos', ascending=False)
```

```python
import matplotlib.pyplot as plt
```

```python
tabela_ordenada = tabela.sort_values(by='Pontos', ascending=False)
plt.bar(tabela_ordenada.index, tabela_ordenada['Pontos'])
plt.xticks(rotation=45, ha='right')
plt.title('Classificação da Premier League 24/25')
plt.ylabel('Pontos')
plt.tight_layout()
```
